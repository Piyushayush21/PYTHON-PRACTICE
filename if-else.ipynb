# LOOPS

A loop is used when we want to repeat some work again and again automatically

Instead of writing the same code many times, we use loops

A loop is a programming structure that:

repeats code multiple times

until a condition becomes false

or until all items are completed

### TYPES OF LOOP 

Python mainly has:

for loop and 
while loop

# FOR LOOP 

````python
for i in range(5):
    print(i)
````

### understanding range 
range(start, stop, step)

````python
for a in range(1, 10, 2):
    print(a)   

    # start from 1
    #stop before 10
    #jump by 2 or step 
````

#### example 2 [table of 2]

````python
for i in range(1, 11):
    print(2 * i)
````

##### LOOPS ON LIST

````python
fruits = ["apple", "banana", "mango"]

for item in fruits:
    print(item)
````

# WHILE LOOP 

Used when we do NOT know exact repetitions

Loop runs until condition becomes False

````python
i = 1

while i <= 5:
    print(i)
    i += 1
````

```text
Start
  ↓
i = 1
  ↓
Check i <= 5
  ↓
True
  ↓
Print i
  ↓
Increase i by 1
  ↓
Go back to condition
  ↓
Repeat until False
```

## Difference Between FOR and WHILE

| FOR LOOP            | WHILE LOOP           |
| ------------------- | -------------------- |
| Fixed repetitions   | Unknown repetitions  |
| Easier              | More flexible        |
| Used with sequences | Used with conditions |


# Loop Control Statements 

BREAK

Stops loop immediately

````python
for i in range(10):

    if i == 5:
        break

    print(i)
````

```text
Loop running
    ↓
Condition true?
    ↓
break executed
    ↓
Loop ends immediately
```

CONTINUE

Skips current iteration.

````python
for i in range(5):

    if i == 2:
        continue

    print(i)

    #2 is skipped
````

PASS

Does nothing.

Used as placeholder.

````python
for i in range(5):

    if i == 4:
        pass

    print(i)
````

```text
Condition True
     ↓
pass executes
     ↓
Nothing happens
     ↓
Program continues normally
```

# Nested Loops

Loop inside another loop

````python
for i in range(3):

    for j in range(5):
        print(i)
````

Workflow of Nested Loop

Outer loop runs once:

inner loop completes fully

Then outer loop moves next.

if-else is a decision-making statement in Python.

It helps the program choose between different actions based on a condition

A nested if means:
An if statement inside another if statement.
Used when one condition depends on another condition.

# IF 

Used to check a condition.

If condition is True, code runs.

````python
# EXAMPLE 
age = 20

if age >= 18:
    print("Adult")
````

```TEXT
Check condition
      ↓
If True → run code
If False → skip code

# ELSE

Used when there are two choices

````python
age = 16

if age >= 18:
    print("Adult")

else:
    print("Minor")
````

# ELIF

Used when there are multiple conditions

````python
marks = 75

if marks >= 90:
    print("Grade A")

elif marks >= 70:
    print("Grade B")

elif marks >= 50:
    print("Grade C")

else:
    print("Fail")
````

# DIFFERNCES

| Statement      | Purpose             |
| -------------- | ------------------- |
| `if`           | Check one condition |
| `if-else`      | Two choices         |
| `if-elif-else` | Multiple choices    |
