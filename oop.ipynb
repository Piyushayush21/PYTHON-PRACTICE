### USING OOP CREATING STUDENT RECORDS  

```python
class Student:
    def __init__(self, name , grade , sports , location):
        self.name = name 
        self.grade = grade  
        self.sports = sports
        self.location = location  ## asigning value 

    def student_details(self):
        print(f"{self.name} is in class {self.grade} and play {self.sports} and he lives in {self.location}")    
```

```python
# assigning value
location = "delhi"
location = "mumbai"
```

```python
Student1 = Student('piyush' , 12 , 'yes' , 'delhi')
# print(Student1.name, Student1.grade , Student1.sports)
```

```python
Student2 = Student('piyushiiii' , 12 , 'no' , 'mumbai')
# print(Student2.name, Student2.grade , Student2.sports) 
```

```python
Student1.student_details()
```

```python
Student2.student_details()
```
