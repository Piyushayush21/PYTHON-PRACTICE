# IF-ELSE STATEMENT

```python
# if : -- executes if the condition is true 
# if-else : --- executes if true, another is false 
# if-elif-else : --- checkes multiple condition in sequence 
```

```python
q = int(input("ENTER YOUR NO."))
if q > 0:
    print("POSITIVE")
elif q < 0:
    print("NEGETIVE")
else:
    print("zero")
```

```python
# even odd check question #
q = int(input("CHECK ODD OR EVEN"))
if q%2 == 0:
    print("EVEN")
else:
    print("ODD")
```

```python
# string ko for loop se reverse kro 'pyhton'
# Method 1: Reverse Using Index (Best Logical Way)
s = "PYTHON"

for i in range(len(s)-1, -1, -1):
    print(s[i]) 
### Explanation:
#len(s)-1 → Last index (5)
#-1 → Stop (exclude)
#-1 → Step backward
```

```python
# Method 2: Using Empty String (Interview Friendly)
s = "PYTHON"
r = ""

for char in s:
    r = char + r

print(r)                 # explanation// Har new character ko aage add kar rahe hain Isliye string reverse hoti ja rahi hai
```

```python
# Method 3: Using While Loop (Extra Knowledge)
s = "PYTHON"
i = len(s) - 1 

while i >= 0:
    print(s[i])
    i -= 1
```
