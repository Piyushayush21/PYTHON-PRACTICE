Exception handling means:

Handling errors in a program so the program does not stop suddenly.

When Python finds an error during execution, it creates an exception.

```text
Exception handling is used to:

Prevent program crash
Handle errors safely
Show proper error messages
Make programs user-friendly
Continue execution even after errors
```

### Main Terminologies in Exception Handling

```text
| Keyword   | Purpose                   |
| --------- | ------------------------- |
| `try`     | Code that may cause error |
| `except`  | Handles the error         |
| `else`    | Runs if no error occurs   |
| `finally` | Runs always               |
| `raise`   | Creates error manually    |

```

```text
| Exception           | Meaning            |
| ------------------- | ------------------ |
| `ZeroDivisionError` | Division by zero   |
| `ValueError`        | Wrong input value  |
| `TypeError`         | Wrong data type    |
| `IndexError`        | Wrong list index   |
| `NameError`         | Variable not found |

```

Workflow of try and except
```text
TRY BLOCK RUNS
      ↓
ERROR FOUND?
   ↓        ↓
 YES        NO
 ↓           ↓
EXCEPT     NORMAL EXECUTION
RUNS
```

````python
try:
    print(10 / 4)

except ZeroDivisionError:
    print("CANNOT DIVIDE BY ZERO")
else:
    print("your answer")
````

````python
try:
    print(10 / 0)

except ZeroDivisionError:
    print("CANNOT DIVIDE BY ZERO")
````

### value error
Occurs when correct data type is expected but wrong value is given.

example of value error 

````python
try:
    age = int(input("ENTER AGE: "))

except ValueError:
    print("ONLY NUMBER ALLOWED")
````

````python
try:
    age = int(input("ENTER AGE: "))

except ValueError:
    print("ONLY NUMBER ALLOWED")
else:
    print("age" , age)
````

### type error 
Occurs when operation happens between incompatible data types

Example

You cannot add string and integer directly

````python
name = "PIYUSH"

age = 20

print(name + age)
#print(name , age)
````

````python
try:
    name = "PIYUSH"

    age = 20

    print(name + age)

except TypeError:
    print("CANNOT ADD STRING AND INTEGER")
````

```text
name = string
age = integer
      ↓
Python tries addition
      ↓
Different types found
      ↓
TypeError occurs
```

### index error 
Occurs when list index does not exist

````python
nums = [10, 20, 30]

print(nums[9])
#print(nums[1])
````

````python
try:
    nums = [10, 20, 30]

    print(nums[5])

except IndexError:
    print("INDEX DOES NOT EXIST")
````

### NameError 
Occurs when variable is not defined

````python
print(memo)
````

````python
try:
    print(memo)

except NameError:
    print("VARIABLE NOT FOUND")
````

````python
try:
    num = int(input("ENTER NUMBER: "))

    numbers = [10, 20, 30]

    print(numbers[5])

    print(10 / num)

except ZeroDivisionError:
    print("ZERO NOT ALLOWED")

except ValueError:
    print("ENTER ONLY NUMBER")

except IndexError:
    print("WRONG INDEX")

except TypeError:
    print("TYPE ERROR")

except NameError:
    print("VARIABLE NOT FOUND")
````
