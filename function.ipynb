# recursion 

Recursion means:

A function calling itself repeatedly until a stopping condition is met

Simple Structure of Recursion

def function_name():

    # base case

    function_name()



````python
def hello():

    print("Hi")
````

````python
hello()
````

```text
| Term               | Meaning                    |
| ------------------ | -------------------------- |
| Recursive Function | Function that calls itself |
| Base Case          | Stopping condition         |
| Recursive Call     | Function calling itself    |

```

````python
# Print numbers from 1 to 10
def no():
    print(no in range(1,11))
````

````python
no()
````
