# FUNCTION 
#### a function is a block of reusable code that performs a specific task. Instead of writing the same code again and again, you can put it inside a function and call it whenever needed

```python
def greet(name):
    return "Hello " + name
print(greet("Alice"))
```

```python
m = 1
n = 3
print(n+m)
```

```python
def cal_sum(a , b):
    add = a + b
    print(add)
    
```

```python
cal_sum(10 , 10)
```

```python
cal_sum(450,50)
```

```python
# write a function to print lenght of list 
def li_len(list):
    print(len(list))

list = [10, 20, 30, 40, 50]
li_len(list)
```

```python
def o_e(n):
    if n%2 ==0:
        print("EVEN")
    else:
        print("ODD")
```

```python
o_e(9)
```

```python
o_e(8)
```

```python
def table(n):
    for i in range(1,11):
        print(n , "x" , i , "=" , n * i)
    print(table)
```
