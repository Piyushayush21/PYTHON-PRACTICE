# DICTIONARY 

A dictionary is a mutable data structure in Python that stores data in the form of key-value pairs

| Property         | Dictionary |
| ---------------- | ---------- |
| Ordered          | ✅          |
| Mutable          | ✅          |
| Duplicate Keys   | ❌          |
| Duplicate Values | ✅          |


```python
student = {
    "name": "Piyush",
    "age": 21
}
print(student)
```

```python
print(student["name"])
```

```python
print(student.get("age"))  # USING GET 
```

```python
student["city"] = "Delhi"
print(student) # ADDING ELEMNT 
```

```python
student["age"] = 22
print(student) # UPDATING VALUE 
```

### Dictionary Methods 

```python
student = {
    "name":"Piyush",
    "age":21
}
print(student.keys())
```

```python
student = {
    "name":"Piyush",
    "age":21
}
print(student.values())
```

```python
print(student.items())
```

```python
student.update({"city":"Delhi"})
print(student)
```

```python
for i in student:
    print(i) # loop through keys 
```

```python
for key, value in student.items():
    print(key, value) # loop thrgh key,values 
```

```python
students = {
    101: {"name":"Piyush", "age":21},
    102: {"name":"Rahul", "age":22}
}
print(students[101]["name"])   #nested dictionary 
```

| Feature  | Dictionary | List          |
| -------- | ---------- | ------------- |
| Stores   | key:value  | values only   |
| Indexing | keys       | numeric index |
| Search   | O(1)       | O(n)          |
